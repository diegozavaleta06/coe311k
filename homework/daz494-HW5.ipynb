```python
import numpy as np
X = np.arange(0, 1.01, 0.01)
X.sort()
random_shift = np.random.uniform(0, 1, len(X))
mtrue = 7
btrue = 3
Y = (mtrue*X+btrue) + random_shift

n = len(X)
sum_x = np.sum(X)
sum_y = np.sum(Y)
sum_x2 = np.sum(X**2)
sum_xy = np.sum(X * Y)
m = (n*sum_xy - sum_x*sum_y) / (n * sum_x2 - (sum_x * sum_x))
c = (sum_y - m * sum_x) / (n) 
y_pred = m * X + c
residuals = Y - y_pred


A = np.vstack([X, np.ones(len(X))]).T
m2, c2 = np.linalg.lstsq(A, Y, rcond=None)[0]
abs_residuals = np.abs(residuals)
max_idx = np.argmax(abs_residuals)
min_idx = np.argmin(abs_residuals)

print(f"Calculated Slope: {m}, Calculated Intercept: {c}")
import matplotlib.pyplot as plt

plt.scatter(X, Y)
plt.plot(X,m*X + c, 'r')
plt.plot(X,m2*X + c2, 'c')



print(f"numpy slope: {m2}, Intercept: {c2}")
print(f"biggest residual: {residuals[max_idx]} at X = {X[max_idx]}")
print(f"smallest residual: {residuals[min_idx]} at X = {X[min_idx]}")

plt.show()
```

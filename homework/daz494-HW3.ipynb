```python
import numpy
def factorial (x):
    if x == 0 or x == 1:
        return 1
    upval = x
    for z in range(1,x):
        upval = upval * (x-z)
    return(upval)



def sin_number(terms):
    listx = []
    for k in numpy.arange(0,10,0.1):
        flipper = 2
        expy = 1
        partsum = 0
        for i in range(0,terms):
           
            sumend = (k**expy) / factorial(expy)
            if flipper%2 == 0:
                partsum = partsum +  sumend
            elif flipper%2 == 1:
                partsum = partsum - sumend   
            expy += 2
            flipper += 1
        real_sin = numpy.sin(k)
        abs_error = abs(real_sin - partsum)
        listx.append(abs_error)
    return(listx)
   
```

```python
import matplotlib.pyplot as plt
x = numpy.arange(0, 10, 0.1)

list1 = sin_number(1)
list2 = sin_number(2)
list3 = sin_number(3)
list4 = sin_number(4)

plt.plot(x,list1,"r", label = "1term")
plt.plot(x,list2,"b", label = "2term")
plt.plot(x,list3,"g", label = "3term")
plt.plot(x,list4,"c", label = "4term")


plt.legend()
plt.xlabel('Values of x in function') 
plt.ylabel('Absolute error') 
plt.title('Taylor Series for Sin(x)') 
plt.show()
```

```python
'''
For small values of x, such as anything below 5, all number of terms seem to peform reasonably well 
as their absolute error isn't too much. However, once we get to larger x values, the error is worse since we 
are farther from zero which is the center. And having more terms at larger x values can actually hurt the accuracy 
because the values that are being subtracted/added are larger in absolute value. So more terms can actually 
increase the truncation error. 

'''
```

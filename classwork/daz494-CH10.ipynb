```python
import numpy as np
''' 
THIS IS PART A OF THE CHALLENGE
'''
X = np.arange(0, 1.01, 0.01)
random_shift = np.random.uniform(-1, 1, len(X))
Y = (12*X+2) + random_shift
A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
import matplotlib.pyplot as plt
y_pred_linear = m * X + c
plt.scatter(X, Y)
plt.plot(X,m*X + c, 'r', label = f'Least Squares method')


#MSE and R-Values for least squares method. 
mse1 = np.mean((Y - y_pred_linear)**2)
print(f'Mean Squared Error for least squares: {mse1:.4f}')
# R-squared value
ss_total1 = np.sum((Y - np.mean(Y))**2)
ss_res1 = np.sum((Y - y_pred_linear)**2)
r_squared1 = 1 - (ss_res1 / ss_total1)
print(f'R-squared for least squares: {r_squared1:.4f}\n')




degree =  3 # Degree of the polynomial
coefficients = np.polyfit(X, Y, degree)
polynomial = np.poly1d(coefficients)
y_pred = polynomial(X)
plt.plot(X, y_pred, color='green', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()

# Mean Squared Error (MSE)
mse = np.mean((Y - y_pred)**2)
print(f'Mean Squared Error for polynomial: {mse:.4f}')
# R-squared value
ss_total = np.sum((Y - np.mean(Y))**2)
ss_res = np.sum((Y - y_pred)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared for polynomial: {r_squared:.4f}')
plt.show()

if r_squared1 > r_squared:
    print("The least squares approach was better")
else:
    print("The polynomial fit was better")
```

```python
import numpy as np
''' 
THIS IS PART C OF THE CHALLENGE
'''
X = np.arange(0, 1.01, 0.01)
random_shift = np.random.uniform(-1, 1, len(X))
Y = (12*X+2) + random_shift
A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
import matplotlib.pyplot as plt
y_pred_linear = m * X + c
plt.scatter(X, Y)
plt.plot(X,m*X + c, 'r', label = f'Least Squares method')


#MSE and R-Values for least squares method. 
mse1 = np.mean((Y - y_pred_linear)**2)
print(f'Mean Squared Error for least squares: {mse1:.4f}')
# R-squared value
ss_total1 = np.sum((Y - np.mean(Y))**2)
ss_res1 = np.sum((Y - y_pred_linear)**2)
r_squared1 = 1 - (ss_res1 / ss_total1)
print(f'R-squared for least squares: {r_squared1:.4f}\n')


degree = 3
V = np.vander(X, N=degree+1, increasing=True)
VT_V = np.dot(V.T, V)
VT_y = np.dot(V.T, Y)
coefficients = np.linalg.solve(VT_V, VT_y)
polynomial = np.poly1d(coefficients[::-1])
y_pred = polynomial(X)
plt.plot(X, y_pred, color='green', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()

# Mean Squared Error (MSE)
mse = np.mean((Y - y_pred)**2)
print(f'Mean Squared Error for polynomial: {mse:.4f}')
# R-squared value
ss_total = np.sum((Y - np.mean(Y))**2)
ss_res = np.sum((Y - y_pred)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared for polynomial: {r_squared:.4f}')
plt.show()

if r_squared1 > r_squared:
    print("The least squares approach was better")
else:
    print("The polynomial fit was better")
```

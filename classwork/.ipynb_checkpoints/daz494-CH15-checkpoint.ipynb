```python
import numpy as np
import matplotlib.pyplot as plt
yfirstrun = []
bool1 = False
multiplier = 1
rangestep = np.arange(0.01,1.01,0.05)
for h_val in rangestep:
    # Define parameters
    h = h_val
    x_start, x_end = 0, 2
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    # Initial condition
    y_values[0] = 1 
    def f(x, y):
        return -2 * y
    # Implement Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    # Plot the results
    plt.plot(x_values, y_values)
    if bool1 == False:
        yfirstrun = y_values.copy()
        bool1 = True
    multiplier = int(round(h / 0.01))
    error = abs(y_values[i] - yfirstrun[i*multiplier])
    total_error = 0

    for j in range(len(y_values)):
        point_error = abs(y_values[j] - yfirstrun[j * multiplier])
        total_error += point_error
        print(f"Point {j}: Error = {point_error:.5f}")

    print(f"TOTAL ACCUMULATED ERROR: {total_error:.5f}")


plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2y using Euler's Method")
plt.show()
```

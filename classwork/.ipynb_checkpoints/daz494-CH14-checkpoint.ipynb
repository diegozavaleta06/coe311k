```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return np.sin(x)
a = 0           
b = 2 * np.pi  
N = 65          
x = np.linspace(a, b, N)



# Step size
h = (b - a) / (N - 1)
# Central Difference Approximation for the derivative
df_dx = (f(x + h) - f(x - h)) / (2 * h)




# Forward Difference Approximation
df_dx2 = (f(x + h) - f(x)) / (h)


# Backward Difference Approximation
df_dx3 = (f(x) - f(x-h)) / (h)


# Plot the approximation and the true derivative
# Plot the approximation and the true derivative
plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx, 'o--',label='Central Difference Approximation')
plt.plot(x, df_dx2, 'o--',label='Forward Difference Approximation')
plt.plot(x, df_dx3, 'o--',label='Backward Difference Approximation')
plt.grid(True)
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Central/Forward/Backward Difference Approx")
plt.legend()
plt.show()

## Do a canvas of 4 subplots or 4 graphs on one plot comparing cos(x), Forward Difference, Backwards Difference, and Central Difference Methods
```

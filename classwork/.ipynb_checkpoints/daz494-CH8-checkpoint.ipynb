```python
import numpy as np
A = np.array([
	[0.582745, 0.48	, 0.10	, 0.  	, 0.  	], 
 [0.48	, 1.044129, 0.46	, 0.10	, 0.  	],
 [0.10	, 0.46	, 1.10431 , 0.44	, 0.10	],
 [0.  	, 0.10	, 0.44	, 0.963889, 0.42	],
 [0.  	, 0.  	, 0.10	, 0.42	, 0.522565]], dtype=float)
def diag_dominal(matrix):
    checker = 0
    n = len(matrix)
    for i in range(n):
        sum = 0
        for j in range(n):
            if i != j:
                val = abs(matrix[i,j])
                sum += val
        if (abs(matrix[i,i]) <= sum):
            checker +=1
        
        
    if checker < 1:
        return True
    else:
        return False
print(A)
print(f"Is the matrix diagonally dominant? {diag_dominal(A)}")
```

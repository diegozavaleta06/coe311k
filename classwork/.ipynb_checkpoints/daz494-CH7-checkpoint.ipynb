```python
import numpy as np
import time
def jacobi(A, b, x0, tol, max_iterations):
    timestart = time.time()
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            timeend = time.time()
            timediff = timeend - timestart
            return x_new, k, timediff
        x = x_new
    timeend = time.time()
    timediff = timeend - timestart
    return x, max_iterations, timediff

# Example usage
A = np.array([
[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)

b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

C = np.array([[1, 1], [1, 1.0001]])
d = np.array([1, 1])

E = np.array([[1, 1], [1, 1.0001]])
f = np.array([1, 1.0001])


x0_1 = np.zeros_like(b, dtype=float)
x0_2 = np.zeros_like(d, dtype=float)
x0_3 = np.zeros_like(f, dtype=float)

tol = 1e-6
max_iterations = 100

solution, iterations, timediff = jacobi(A, b, x0_1, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"It took this long to solve: {timediff}")
print("")


solution2, iterations2, timediff2 = jacobi(C, d, x0_2, tol, max_iterations)
print("This is for second matrix set")
print(f"Solution: {solution2}")
print(f"Iterations: {iterations2}")
print(f"It took this long to solve: {timediff2}")
print("")


solution3, iterations3, timediff3 = jacobi(E, f, x0_3, tol, max_iterations)
print("This is for thid matrix set")
print(f"Solution: {solution3}")
print(f"Iterations: {iterations3}")
print(f"It took this long to solve: {timediff3}")
```

```python
import numpy as np
import time
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    timestart = time.time()

    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            timeend = time.time()
            timediff = timeend - timestart
            return x_new, k, timediff
        x = x_new
    timeend = time.time()
    timediff = timeend - timestart
    return x, max_iterations, timediff

# Example usage
A = np.array([
[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)

b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

C = np.array([[1, 1], [1, 1.0001]])
d = np.array([1, 1])

E = np.array([[1, 1], [1, 1.0001]])
f = np.array([1, 1.0001])

x0_1 = np.zeros_like(b, dtype=float)
x0_2 = np.zeros_like(d, dtype=float)
x0_3 = np.zeros_like(f, dtype=float)
tol = 1e-6
max_iterations = 100

solution, iterations, timediff = gauss_seidel(A, b, x0_1, tol, max_iterations)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"It took this long to solve: {timediff}")
print("")


solution2, iterations2, timediff2 = gauss_seidel(C, d, x0_2, tol, max_iterations)
print("This is for second matrix set")
print(f"Solution: {solution2}")
print(f"Iterations: {iterations2}")
print(f"It took this long to solve: {timediff2}")
print("")


solution3, iterations3, timediff3 = gauss_seidel(E, f, x0_3, tol, max_iterations)
print("This is for thid matrix set")
print(f"Solution: {solution3}")
print(f"Iterations: {iterations3}")
print(f"It took this long to solve: {timediff3}")

```

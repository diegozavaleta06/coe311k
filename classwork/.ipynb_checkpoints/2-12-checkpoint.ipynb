# Linear algebra in python

```python
import numpy as np
```

```python
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
```

```python
print(v1*v2)
```

```python
print(np.dot(v1,v2))
```

```python
print(np.cross(v1,v2))
```

```python
magnitude = np.linalg.norm(v2)
```

```python
print(magnitude)
```

```python
A = np.array([[1,2,3],[4,5,6]])
print(A)
print()
B = np.array([[7, 8, 9], [10, 11, 12]])
print(B)
```

```python
C = np.cross(A,B)
print(C)
```

```python
A = np.array([[2,3],[4,-1]])
B = np.array([5,3])
x = np.linalg.solve(A,B)
```

```python
print(x)
```

```python
A = np.array([[2,1],[1,2]])
e_value, e_vectors = np.linalg.eig(A)
```

```python
print(e_value)
print(e_vectors)
```

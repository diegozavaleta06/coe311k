```python
import numpy as np
# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([3.2, 5.8, 3.6, 6.5, 5.1])
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

print("X:")
print(X)

print("Y:")
print(Y)

print("A:")
print(A)

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
import matplotlib.pyplot as plt

plt.scatter(X, Y)
plt.plot(X,m*X + c)
plt.show()
```

```python
np.random.shuffle(Y)
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
import matplotlib.pyplot as plt

plt.scatter(X, Y)
plt.plot(X,m*X + c)
plt.show()
```

```python
X = np.arange(0, 1.01, 0.01)
random_shift = np.random.uniform(-1, 1, len(X))
Y = (12*X+2) + random_shift
A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
import matplotlib.pyplot as plt

plt.scatter(X, Y)
plt.plot(X,m*X + c, 'r')
plt.show()
```

```python
import time
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

import numpy as np

n = 3  
matrix_pairs = []

for i in range(8):
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)
    
    matrix_pairs.append((A, B))
    
    print(f"Pair {i+1} generated: {n}x{n}")
    n *= 2  

counter = 1
naive_list = []
for i in matrix_pairs:
    A, B = i
    start_time = time.time()
    result = naive_matrix_multiplication(A,B)
    end_time = time.time()
    duration = end_time - start_time
    print(f"The naive method took {duration} seconds for iteration {counter}.")
    naive_list.append(duration)
    counter += 1
print("\n")
counter = 1
np_list = []
for i in matrix_pairs:
    A, B = i
    start_time = time.time()
    result = np.dot(A,B)
    end_time = time.time()
    duration = end_time - start_time
    print(f"The numpy method took {duration} seconds for iteration {counter}.")
    np_list.append(duration)
    counter += 1    
    
import matplotlib.pyplot as plt

xlength = [1,2,3,4,5,6,7,8]
plt.plot(xlength,naive_list, label = 'naive')
plt.plot(xlength,np_list, label = 'np')
plt.title('Matrix Multiplication Runtime')
plt.xlabel('Iteration Number')
plt.ylabel('Time')
plt.legend()
plt.show()
```

```python
import numpy as np
A = np.array([
	[0.582745, 0.48	, 0.10	, 0.  	, 0.  	], 
 [0.48	, 1.044129, 0.46	, 0.10	, 0.  	],
 [0.10	, 0.46	, 1.10431 , 0.44	, 0.10	],
 [0.  	, 0.10	, 0.44	, 0.963889, 0.42	],
 [0.  	, 0.  	, 0.10	, 0.42	, 0.522565]], dtype=float)
def diag_dominal(matrix):
    checker = 0
    c = 0
    d = 0
    n = len(matrix)
    for i in range(n):
        sum = 0
        for j in range(n):
            if (c!=i) & (d!=j):
                val = matrix[i,j]
                sum += val
        if (matrix[c,d] < sum):
            checker +=1
        c += 1
        d += 1
        
    if checker < 1:
        print("It is diagonally dominant")
    else:
        print("It isn't diagonally dominant")
        
diag_dominal(A)
```
